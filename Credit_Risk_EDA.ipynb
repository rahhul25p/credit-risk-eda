```python
import pandas as pd

df = pd.read_csv('/content/cleaned_data.csv')
```

```python
df.groupby('loan_grade')['loan_amnt'].mean()
```

```python
df.groupby('loan_grade')['person_income'].mean()
```

```python
df.groupby('loan_intent')['loan_status'].mean()*100
```

```python
df.groupby('person_home_ownership')['loan_status'].mean()*100
```

```python
df['loan_intent'].value_counts()
```

```python
df['person_home_ownership'].value_counts()
```

```python
pd.crosstab(df['person_home_ownership'], df['loan_status'])
```

```python
age_group = pd.cut(
    df['person_age'],
    bins=[0, 25, 35, 50, 1000],
    labels=['25 or younger', '26–35', '36–50', 'Over 50']
)

age_group.value_counts().sort_index()
```

```python
pd.pivot_table(
    df,
    values='loan_status',
    index='loan_intent',
    columns='person_home_ownership',
    aggfunc='mean'
) * 100
```

```python
(df.groupby('loan_grade')['loan_status'].mean() * 100).plot(
    kind='bar',
    title='Default Rate by Loan Grade',
    xlabel='Loan Grade',
    ylabel='Default Rate (%)'
)
```

```python
(df.groupby('loan_intent')['loan_status'].mean()*100).sort_values().plot(kind='bar',title='Default Rate by Loan Intent',xlabel='Loan Intent',ylabel='Default Rate(%)')
```

```python
df.groupby('loan_status')['loan_amnt'].mean().plot(
    kind='bar',
    title='Average Loan Amount by Loan Status',
    xlabel='Loan Status (0 = Non-defaulted, 1 = Defaulted)',
    ylabel='Average Loan Amount'
)
```

```python
(df.groupby('person_home_ownership')['loan_status'].mean() * 100).sort_values().plot(
    kind='bar',
    title='Default Rate by Home Ownership',
    xlabel='Home Ownership',
    ylabel='Default Rate (%)'
)
```
